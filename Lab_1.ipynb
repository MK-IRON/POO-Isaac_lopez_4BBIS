```python
class headphones:
    def __init__(self, microphone, frequency, style, mark):
        self.microphone = microphone
        self.frequency = frequency
        self.style = style
        self.mark = mark

    def play(self):
        print(f"The {self.mark} are playing music")

    def describe(self):
        if self.microphone == True:
            print(f"The {self.mark} headphones includes a microphone")
        else:
            print(f"The {self.mark} headphones dont include a microphone")
        print(f"The {self.mark} headphones have a frequency of {self.frequency}")
        print(f"The {self.mark} headphones are {self.style}")

razerHedphones = headphones(True,"20k/hz","on-ear","razer")
hyperxHeadphones = headphones(False, "18k/hz", "in-ear","hyperx")

razerHedphones.describe()
razerHedphones.play()

print("=====")

hyperxHeadphones.describe()
hyperxHeadphones.play()
```

# Lab 1: Bank account Class

create a Python program that models a bank account using
Objected-Oriented-Programming

Follow these requierements:

  1.Create a class called bankAccount
  
  2. Add a construcor using init that receives the account holder´s name. and the Initial balance

  3.Create the following attributes
    holder -> stores the account holders name
    __balance -> stores the account balances
    
  4.Create a method called deposit()
    that receives an amount of money adds the amount of monety to the account balance and does not allow negative or zero amounts

```python
class bankAccount:
  def __init__(self, holder, balance):
    self.holder = holder
    self.balance = balance

  def deposit(self, money):
    if money <= 0:
      print("Invalid amount")
      return
    else:
      self.balance += money

  def withdraw(self, money):
    if money>=self.balance:
      print("Invalid operation")
    else:
      self.balance -= money

  def checkbalance(self):
    print(f"Balance: {self.balance}")

account1 = bankAccount("Juan Perez" ,5000)
account2 = bankAccount("Maria López",3000)

account1.checkbalance()
account1.deposit(1000)
account1.checkbalance()
account1.withdraw(500)
account1.checkbalance()

print("=======")

account2.checkbalance()
account2.deposit(1000)
account2.checkbalance()
account2.withdraw(500)
account2.checkbalance()
```
